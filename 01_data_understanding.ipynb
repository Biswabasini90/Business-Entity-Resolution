```python
singleton_count = (
    ground_truth["matched_entity_ids"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print("Number of singletons:", singleton_count)
```

```python
def count_matches(value):
    if pd.isna(value) or str(value).strip() == "":
        return 0
    
    return len(
        str(value).split(",")
    )
```

```python
ground_truth["match_count"] = (
    ground_truth["matched_entity_ids"]
    .apply(count_matches)
)
```

```python
missing_s1 = (
    set(ground_truth["source1_entity_id"])
    -
    set(train_s1["entity_id"])
)

print(
    "Ground-truth S1 IDs not found in Source 1:",
    len(missing_s1)
)
```

```python
s2_ids = set(train_s2["entity_id"])
s3_ids = set(train_s3["entity_id"])

all_target_ids = s2_ids | s3_ids
```

```python
def get_matched_ids(value):
    if pd.isna(value) or str(value).strip() == "":
        return []
    
    return [
        x.strip()
        for x in str(value).split(",")
        if x.strip()
    ]
```

```python
ground_truth["matched_id_list"] = (
    ground_truth["matched_entity_ids"]
    .apply(get_matched_ids)
)
```

```python
invalid_ids = []

for ids in ground_truth["matched_id_list"]:
    for entity_id in ids:
        if entity_id not in all_target_ids:
            invalid_ids.append(entity_id)

print("Invalid matched IDs:", len(invalid_ids))
```
